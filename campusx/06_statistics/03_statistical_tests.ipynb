# Statistical Analysis

## 1. Chi-square Test

```python
# import libraries
import pandas as pd
import numpy as np
import seaborn as sns
from scipy import stats
```

```python
df = sns.load_dataset('titanic')
df.head()
```

`Null Hypothesis:` There is no significant association gender ('sex') and survival ('survived') on the Titanic. This means any observed differences in survival differences in Survial rates between genders in the dataset are due to chance and not due to an underlying relationship.

`Alternative Hypothesis:` There is a significant association between gender ('Sex') and survial ('Survived') on the Titanic. This implies that the differences in survival rates are not just due to chance but are influenced by the passenger's gender.

```python
# Create a contingency table
contingency_table = pd.crosstab(df['sex'], df['survived'])
contingency_table
```

```python
# chi-square test
chi2, p, dof, expected = stats.chi2_contingency(contingency_table)

# print the results
print('Chi-square statistics:', chi2)
print('p-value:', p)
print('Degrees of freedom:',dof)
print(f"Expected:\n {expected}")
```

As the p_value in this test is `P-value: 1.1973570627755645e-58` which is far less than 0.05, we reject the null hypothesis.

# shapiro wilk test on normal distribution gaussian distribution
# Null Hypothesis: the data is normally distributed.
# Alternative Hypothesis: data is not normally distributed.

```python
# shapiro wilk test
statistic, p = stats.shapiro(df['fare'])
# print the results using if else
if p > 0.05:
    print(f'p-value: {p}, \nSample looks Gaussian/Normally Distributed (fail to reject H0)')
else:
    print(f'p-value: {p}, \nSample does not look Gaussian/Normally Distributed (reject H0)')
```

```python
sns.histplot(data=df, x='fare', kde=True)
```

```python
# kolmogrov test
age = df['age'].dropna() # remove null values

# mean & std
mean = age.mean()
std = age.std()

# KS test
statistic, p_value = stats.kstest(age, 'norm', args=(mean, std))

print("KS Statistic:", statistic)
print("p-value:", p_value)
```

# t-test

1. one-sample t-test

```python
import scipy.stats as stats

# sample data
x = [1, 2, 3, 4, 5]

# known population mean
mu = 5
# perform one sample t-test
t_statistic, p = stats.ttest_1samp(x, mu)

# print results
print("t-statistic:", t_statistic)
print("p-value:", p)

# print the result using if else
if p > 0.05:
    print(f'p-value: {p}, Sample mean is equal to population mean (fail to reject H0)')
else:
    print(f"p-value: {p}, Sample mean is not equat ot pupulation mean (reject H0)")
```

2. Two Sample t-test (independent)

```python
from scipy import stats
group1 = [2.3, 3.4, 4.5, 2.3, 3.4]
group2 = [1.2, 2.2, 3.2, 2.2, 2.3]

# perform the independent two-sample t-test
t_stat, p = stats.ttest_ind(group1, group2, equal_var=False)

print("t-statistic:", t_stat)
print("p-value:", p)

if p > 0.05:
    print(f"p-value: {p}, group1 mean is equal to group2 mean (fail to reject H0)")
else:
    print(f"p-value: {p}, group1 mean is not equal to group2 mean (reject H0)")
```

3. Paired Sample t-test

```python
import scipy.stats as stats

# sample data
before = [2, 3, 4, 5, 6]
after = [3, 4, 5, 6, 7]

# perform paired sample t-test
t_stat, p_val = stats.ttest_rel(before, after)

print("t-statistic:", t_stat)
print("p-value:", p_val)

# print the result using if else
if p_val > 0.05:
    print(f"p-value: {p_val}, before mean is equal to after mean (fail to reject H0)")
else:
    print(f"p-value: {p_val}, before mean is not equal to after mean (reject H0)")
```

# Test for Homogeniety

## 1. Levene's Test

```python
import scipy.stats

# Sample data
group1 = [2.3, 3.4, 4.5, 2.3, 3.4]
group2 = [1.2, 2.2, 3.2, 2.2, 2.2]

# Perform Levene's Test
w_stat, p_val = stats.levene(group1, group2)

print("w_statistic:", w_stat)
print("p-value:", p_val)

if p_val > 0.05:
    print(f"p-value: {p_val}, Variance are equal (fail to reject H0)")
else:
    print(f"p-value: {p_val}, Variance are not equal (reject H0)")
```

## 2. Bartlett's Test

```python
import scipy.stats

# Sample data
group1 = [2.3, 3.4, 4.5, 2.3, 3.4]
group2 = [1.2, 2.2, 3.2, 2.2, 2.3]

# Perform Bartlett's Test
t_stat, p_val = stats.bartlett(group1, group2)

print("t-statistics:", t_stat)
print("p-value:", p_val)

# print the results using if else
if p_val > 0.05:
    print(f"p-value: {p_val}, Variance are equal (fail to reject H0)")
else:
    print(f"p-value: {p_val}, Variance are not equal (reject H0)")
```

# ANOVA
Analysis of Variance

## one-way Anova

```python
import scipy.stats as stats

# Sample data
a = [2.3, 3.4, 4.5, 2.3, 3.4]
b = [11.2, 12.2, 13.2, 12.2, 12.3]
c = [22.3, 23.4, 24.5, 22.3, 23.4]

# perform one-way ANOVA
f_stat, p_val = stats.f_oneway(a, b, c)

# print the results base on if else conditions
if p_val > 0.05:
    print(f"p-value: {p_val}, Means are equal (fail to reject H0)")
else:
    print(f"p-value: {p_val}, Means are not equal (reject H0)")
```
