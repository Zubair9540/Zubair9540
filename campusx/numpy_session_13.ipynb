# Stacking

```python
# horizontal stacking
b = np.arange(12).reshape(3,4)
c = np.arange(12,24).reshape(3,4)
c
```

```python
np.hstack((b,c))
```

```python
# vertical stacking
np.vstack((b,c))
```
