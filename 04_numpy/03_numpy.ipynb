# Matrix or Matrices

```python
# create a 2x3 matrix
a = np.array([[1,2,3],[4,5,6]])
a
```

```python
# create a matrix
b = np.arange(0, 100, 2).reshape(5,10) # array[row_start : row_end , column_start : column_end]
b
```

```python
# slicing a matrix
b[0:3, 3:5]
```

```python
b.min()
```

```python
b.max()
```

```python
b
```

```python
b.max(axis=0) # only for columns
```

```python
b.max(axis=1) # only for rows
```
