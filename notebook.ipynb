```python
import plotly.express as px
import pandas as pd

# Sample data
df = pd.DataFrame({
    "Year": [2015, 2016, 2017, 2018, 2019, 2020],
    "Sales": [100, 120, 150, 170, 160, 200],
    "Profit": [20, 25, 30, 35, 33, 45]
})

# Create figure
fig = px.line(df, x="Year", y=["Sales", "Profit"], title="Sales vs Profit")

# Show in browser (bypasses nbformat/ipython issue)
fig.show(renderer="browser")
```
