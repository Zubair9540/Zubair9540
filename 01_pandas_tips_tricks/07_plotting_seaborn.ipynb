# Seaborn and plotting

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Important things for plotting according to grammar of graphics

1. Data(df)
2. Types of plot?
3. x axis
4. y axis
5. grouping (hue/color) based on categorical variables
6. Size of the plot
7. color pallets of the plot
8. Style of the plot
9. How do you interpret

```python
df = sns.load_dataset('titanic')
```

```python
df.info()
```

```python
# scatterplot
sns.scatterplot(data = df, x='age', y='fare' , hue='sex')
```

```python
# scatter plot size of figure size
plt.figure(figsize=(10,4))
sns.scatterplot(data=df, x = 'age', y = 'fare', hue='sex')
```

```python
# change scatter point colors
plt.Figure(figsize=(8,4))
palette = {"male": "blue", "female": "red"}
sns.scatterplot(data = df, x = 'age', y = 'fare', hue = 'sex', palette=palette)
# labels
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Scatter Plot of Age vs Fare')
# saving the plot under 150dpi, 300dpi, 600dpi
plt.savefig('./outputs/plots/scatter_plot.png', dpi=300)
# saving a plot in vector file such as svg
plt.savefig('./outputs/plots/scatter_plot.svg')
```

```python
# style of a plot
sns.set_theme(style='darkgrid',)
plt.figure(figsize=(8,4))
palette = {'male':'blue', 'female':'red'}
sns.scatterplot(data=df,x='age',y='fare',hue='sex',palette=palette)
plt.xlabel('Age')
plt.ylabel('Fare')
plt.title('Scatter Plot of Age vs Fare')
```

Interpertation:
1. We have Age of passengers at x-axis and fare at y-axis
2. The scatter plot shows the relationship between age and fare of titanic passengers.
3. We further grouped the passenger by their genders such as male (blue points) and female(red points)
4. Two male passengers gave the highest fare in titanic, and they might be traveling first class.
5. The oldest person at titanic is also a male having 80 years of age as shown in the scatter plot. (As shown in Figure.1)

# Box Plot

```python
sns.set_theme(style='darkgrid',font_scale=1,font='Times New Roman')
plt.figure(figsize=(12,6))
palette = {'First':'green', 'Second':'red', 'Third':'blue'}
sns.boxplot(data=df,x='sex',y='age',hue='class',palette=palette)
plt.xlabel('Sex')
plt.ylabel('Age')
plt.title('Scatter Plot Of Sex Vs Age')
# change legend position and title
plt.legend(loc='upper right', title='Class Names')
```

```python
sns.set_theme(style='darkgrid',font_scale=1,font='Times New Roman')
plt.figure(figsize=(12,6))
palette = {'First':'green', 'Second':'red', 'Third':'blue'}
sns.boxenplot(data=df,x='sex',y='age',hue='class',palette=palette)
plt.xlabel('Gender')
plt.ylabel('Age')
plt.title('Boxen Plot Of Age, based on Gender and Class Name of Passengers')
# change legend position and title
plt.legend(loc='upper right', title='Class Names')
# change the x-ticks  names
plt.xticks(ticks=[0,1], labels=["Male", "Female"], rotation=45)
```

```python
# bar plot using seaborn
sns.set_theme(style='darkgrid', font_scale=1, font='Times New Roman')
plt.figure(figsize=(12,6))
palette={'First': 'green', 'Second': 'red', 'Third':'blue'}
sns.barplot(data=df,x='sex',y='age',hue='class',palette=palette,capsize=0.1)
plt.xlabel('Gender')
plt.ylabel('Age')
plt.title('Barpot of Age, Based on Gender and Class Name of Passengers')
plt.legend(loc='upper right',title='Class Name')
plt.xticks(ticks=[0,1], labels=["Male", "Famale"],rotation=0)
```
