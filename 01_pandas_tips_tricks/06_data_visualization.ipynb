# Data Visualization

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# imprt titanic data from seaborn library
df = sns.load_dataset('titanic')
```

```python
df.plot(kind='scatter', x='age', y='fare', color='orange')
```

```python
# change colors using hex codes
df.plot(kind='scatter', x='age', y='fare', color='#13c3d6')
```

```python
df.head()
```

```python
df.groupby('sex')['fare'].mean().plot(kind='bar')
plt.ylabel('Average Fare')
plt.xlabel('Gender')
```

```python
# seaborn plots

# bar plots
sns.scatterplot(data= df, y='fare',x = 'age', hue='sex' )
```
