# Inverse encoding/ Inverse decoding categorical variables

```python
# import libraries
import pandas as pd
import numpy as numpy
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
df.isnull().sum()
```

```python
# impute missing values
df['age'] = df['age'].fillna(df['age'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])
# drop deck column
df = df.drop('deck', axis=1)
```

```python
# encode all categorical variables and object variables

le_sex = LabelEncoder()
le_embarked = LabelEncoder()
le_class = LabelEncoder()
le_who = LabelEncoder()
le_embark_town = LabelEncoder()
le_alive = LabelEncoder()

df['sex'] = le_sex.fit_transform(df['sex'])
df['embarked'] = le_embarked.fit_transform(df['embarked'])
df['class'] = le_class.fit_transform(df['class'])
df['who'] = le_who.fit_transform(df['who'])
df['embark_town'] = le_embark_town.fit_transform(df['embark_town'])
df['alive'] = le_alive.fit_transform(df['alive'])
```

```python
df.head()
```

```python
df['sex'] = le_sex.inverse_transform(df['sex'])
df['embarked'] = le_embarked.inverse_transform(df['embarked'])
df['class'] = le_class.inverse_transform(df['class'])
df['who'] = le_who.inverse_transform(df['who'])
df['embark_town'] = le_embark_town.inverse_transform(df['embark_town'])
df['alive'] = le_alive.inverse_transform(df['alive'])
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder

df = sns.load_dataset('titanic')

cat_columns = ['sex', 'embarked']

encoder = OneHotEncoder(sparse_output=False, drop='first')
encoded_df = pd.DataFrame(encoder.fit_transform(df[cat_columns]))

# concatenate the dataframes
df = pd.concat([df, encoded_df], axis=1)
df.head()
```

```python
# we need to extract the original categories for each encoded column
original_categories = {col: encoder.categories_[i] for i, col in enumerate(cat_columns)}

# manual creation of feature names
feature_names = []
for i, col in enumerate(cat_columns):
    for category in encoder.categories_[1:]: 
        feature_names.append(f"{col}_{category}")

encoded_df = pd.DataFrame(encoded_df, columns=feature_names)
df.head()
```

```python
# ordinal encoder
import pandas as pd
import seaborn as sns
from sklearn.preprocessing import OrdinalEncoder

# 1. Titanic dataset load karna
df = sns.load_dataset('titanic')

# Hum sirf 2 columns lete hain taake samajhna asaan ho
# 'class' (First, Second, Third) aur 'embark_town' (Southampton, etc.)
df_subset = df[['class', 'embark_town']].dropna().copy()

# 2. Ordinal Encoder ko Setup karna
# Hum batate hain ke 'class' ki tarteeb (order) kya honi chahiye
class_order = ['First', 'Second', 'Third']
towns = ['Southampton', 'Cherbourg', 'Queenstown']

encoder = OrdinalEncoder(categories=[class_order, towns])

# 3. Fit aur Transform (Words ko Numbers mein badalna)
encoded_values = encoder.fit_transform(df_subset)

# Encode kiye hue data ko ek naye DataFrame mein dekhna
encoded_df = pd.DataFrame(encoded_values, columns=['class', 'embark_town'])
print("--- Encoded Data (Numbers) ---")
print(encoded_df.head())

# 4. Inverse Transform (Numbers ko wapis Words mein badalna)
original_values = encoder.inverse_transform(encoded_values)

# Wapis asli halat mein dekhna
decoded_df = pd.DataFrame(original_values, columns=['class', 'embark_town'])
print("\n--- Inverse Transformed Data (Back to Words) ---")
print(decoded_df.head())
```
