# 1. Min Max Scaling

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
# sample data
data = {'Value':[10,20,30,40,50]}
df = pd.DataFrame(data)
df.head()
```

```python
# Min-Max Scaling
scaler = MinMaxScaler()
df['Scaled_Value'] = scaler.fit_transform(df[['Value']])
print(df)
```

# 2. Standard Scalar or Z-score normalization

```python

import pandas as pd
from sklearn.preprocessing import StandardScaler

```

```python
# sample data
data = {'value':[60,70,80,90,100]}
df = pd.DataFrame(data)
df.head()
```

```python
# scale the data using standard scaler
scaler = StandardScaler()
df['value_scaled'] = scaler.fit_transform(df[['value']])
df.head()
```

# 3. Robust Scalar

```python
from sklearn.preprocessing import RobustScaler
# sample data with outliers
data = {'value':[10,20,30,1000,50]}
df = pd.DataFrame(data)
df.head()
```

```python
# Robust Scaling
scaler = RobustScaler()
df['value'] = scaler.fit_transform(df[['value']])
print(df)
```

# 4. Logrithmic Scaling / Normalization

```python
import numpy as np
import pandas as pd
# random data with outliers
data = {'value':[10000,20000,30000,1000000,50000]}
df = pd.DataFrame(data)
df.head()
```

```python
# simple log transforms (natural, base2, base10)
df['value_log'] = np.log(df['value'])
df['value_log2'] = np.log2(df['value'])
df['value_log10'] = np.log10(df['value'])
df.head()
```

```python
import seaborn as sns
sns.boxplot(df,x="value")
```
